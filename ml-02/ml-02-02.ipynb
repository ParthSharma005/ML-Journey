```python
df_encode = pd.get_dummies(df, drop_first=True, dtype=int)
df_encode.head()
```

Label Encoding - Assigns distinct numerical values to each category w/o creating dummies

```python
from sklearn.preprocessing import LabelEncoder
```

```python
label = LabelEncoder()
df_label = df.copy()
category_cols = ['model', 'transmission', 'fuelType']
for col in category_cols:
    df_label[col] = label.fit_transform(df[col])

df_label.head()
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
scaler = StandardScaler()
to_scale = ['year', 'mileage', 'tax', 'mpg', 'engineSize']
df_encode[to_scale] = scaler.fit_transform(df_encode[to_scale])
df_encode
```

Unlike One-Hot, all columns of Label Encoded should be scaled (except Target column, i.e. price) 

```python
df_label[['model', 'year', 'transmission', 'mileage', 'fuelType', 'tax',
       'mpg', 'engineSize']] = scaler.fit_transform(df_label[['model', 'year', 'transmission', 'mileage', 'fuelType', 'tax',
       'mpg', 'engineSize']])
df_label
```

#Data Splitting & Model Training 

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error
```

```python
x_encode = df_encode.drop('price', axis=1)
x_label = df_label.drop('price', axis=1)
y = df['price']
```

```python
x_encode_train, x_encode_test, y_train, y_test = train_test_split(x_encode, y, train_size=0.8, random_state=42) 
x_label_train, x_label_test, y_train, y_test = train_test_split(x_label, y, train_size=0.8, random_state=42) 
```

```python
model = LinearRegression()
model.fit(x_encode_train, y_train)
```

```python
y_pred_encode = model.predict(x_encode_test)
```

```python
model.fit(x_label_train, y_train)
```

```python
y_pred_label = model.predict(x_label_test)
```

#Performance Metrics

```python
pl = sns.scatterplot(x=y_test, y=y_pred_encode)
pl.axline((0,0), slope=1, color='red', linestyle='--')
```

```python
pl = sns.scatterplot(x=y_test, y=y_pred_label)
pl.axline((0,0), slope=1, color='red', linestyle='--')
```

```python
r2_encode = r2_score(y_test, y_pred_encode)
r2_encode
```

```python
n = x_encode_test.shape[0]
p = x_encode_test.shape[1]
adr2_encode = 1-(((1-r2_encode)*(n-1))/(n-p-1))
adr2_encode
```

```python
r2_label = r2_score(y_test, y_pred_label)
r2_label
```

```python
p2 = x_label_test.shape[1]
adr2_label = 1-(((1-r2_label)*(n-1))/(n-p-1))
adr2_label
```

Label Encoding returned lower metrics than one-hot but the former utilised less columns than latter. 

Label is better when data categories have ordered relationships (eg., Low, Medium, High = 0,1,2)
